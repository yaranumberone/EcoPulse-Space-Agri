```python
import numpy as np
import matplotlib.pyplot as plt

# 1. محاكاة واستخراج نطاقات القمر الصناعي Sentinel-2 (Band 8: NIR, Band 4: RED, Band 11: SWIR)
np.random.seed(42)
grid_size = (100, 100)

# إنشاء مصفوفات طيفية تمثل حقل زراعي بمناطق جافة ومناطق رطبة
nir = np.random.normal(0.6, 0.1, grid_size)   # Band 8 (Near-Infrared)
red = np.random.normal(0.2, 0.05, grid_size)  # Band 4 (Red)
swir = np.random.normal(0.3, 0.08, grid_size) # Band 11 (Short-Wave Infrared)

# 2. حساب المؤشرات الطيفية للأقمار الصناعية
# NDVI = (NIR - RED) / (NIR + RED)
ndvi = (nir - red) / (nir + red)

# NDWI = (NIR - SWIR) / (NIR + SWIR)
ndwi = (nir - swir) / (nir + swir)

# 3. رسم وتوليد الخرائط الخاصة بالـ Pitch Deck
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# خريطة صحة النبات (NDVI)
im1 = axes[0].imshow(ndvi, cmap='YlGn')
axes[0].set_title('Sentinel-2 Crop Health Index (NDVI)')
plt.colorbar(im1, ax=axes[0], label='NDVI Value')
axes[0].axis('off')

# خريطة رطوبة التربة (NDWI)
im2 = axes[1].imshow(ndwi, cmap='Blues')
axes[1].set_title('Sentinel-2 Soil Moisture Index (NDWI)')
plt.colorbar(im2, ax=axes[1], label='NDWI Value')
axes[1].axis('off')

plt.tight_layout()
plt.savefig('sentinel_satellite_analysis.png', dpi=300)
plt.show()

print("✅ تم استخراج وحساب مؤشرات الأقمار الصناعية بنجاح! تم حفظ الصورة باسم sentinel_satellite_analysis.png")
```
